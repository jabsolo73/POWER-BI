Data Modeling (16–25)

16. What is data modeling?
- Building relationship between tables
17. What is the difference between a Fact table and a Dimension table?
- Fact table stores numerical
- Dimension table stores descriptive
18. What is a Star Schema?
- one Fact table in the center connected directly to multiple Dimension tables.
19. What is a Snowflake Schema?
- similar to a Star Schema, but the Dimension tables are further split into additional related tables.
20. Why is Star Schema preferred in Power BI?
- More speed and easier to understand
21. Explain one-to-one, one-to-many, and many-to-many relationships.
- 1 - 1: One record in Table A matches only one record in Table B
- 1 - many: One record in Table A matches many records in Table B, but those records link back to only one entry in Table A.
- many - many:  Multiple records in Table A can connect to multiple records in Table B.
22. What is the difference between Active and Inactive relationships?
- Active: Used by default for all standard visual filters and calculations. Only one active path is allowed between any two tables. Shown as a solid line.
- Inactive: : Stored in the data model as an alternate path, but do not pass filters on their own. Must be turned on manually inside a measure using a function like USERELATIONSHIP. Shown as a dashed or dotted line.
23. What is Cross Filter Direction?
- It can be selected when creating a relationship
- controls how filters flow between connected data tables: single-direction and both (bi-directional)
24. What is a Date Table, and why is it required?
- It is a seperate table consist of continuous periodic date values each row
- Used for Time intelligence function most
25. What is the difference between Import mode and DirectQuery?
- Import mode: copies and stores data into Power BI’s in-memory storage(cloud)
- Direct Query: leaves database in the source and queries it live(data source)

29. Where are Measures stored in the model?

30. Can a Measure be used as a slicer? Why or why not?
- No, Slicer needs columns to perform the filter where measure is just a value

Filter & Context Functions (37–43)

37. What is Filter Context?
- Filter Context is the collection of filters applied to a calculation. Measures automatically recalculate based on the current filter context.
- Goes model level too and changes at relationship level too
38. What is Row Context?
- works row by row
- can access the value of each column of only the iterated table. Columns in different tables are not affected by the presence of a row context, even if there are relationships linking the tables together.
39. What is the FILTER() function?
- returns only the rows that satisfy a condition.
- FILTER() returns a filtered table based on a condition and is commonly used inside CALCULATE() and iterator functions.
40. What is the CALCULATE() function?
- it is a overall guiding where inside this function expression and filters are given
41. Why is CALCULATE() considered the most powerful DAX function?
- can concat multiple expression and filters
42. What is the difference between ALL() and FILTER()?
- ALL():  tells Power BI to ignore any filters currently applied, commonly used inside CALCULATE() to create denominators for percentages
- Filter(): is an iterator function. It looks at a table, evaluates a specific true/false condition for every single row, and returns only the rows that match.
43. What is the difference between ALL(), ALLEXCEPT(), and REMOVEFILTERS()?
- ALL()	Removes all filters from a table or specified columns and returns all rows
- ALLEXCEPT()	Removes all filters except the specified columns
- REMOVEFILTERS()	Removes filters from the specified table or columns without returning a table for iteration

Logical Functions (44–48)

44. What is the IF() function?
- The IF() function checks a condition. If the condition is TRUE, it returns one value.
- If the condition is FALSE, it returns another value.
45. What is the difference between IF() and SWITCH()?
- IF(): Best for one or two conditions.
- SWITCH(): is cleaner when you have many possible results and easier to read than using nested IF
46. What is the purpose of AND()?
- returns true only if both conditions are true
47. What is the purpose of OR()?
- returns if either one condition is true
48. What is the purpose of NOT()?
- reverses a logical value.
- TRUE becomes FALSE.
- FALSE becomes TRUE.

Date & Time / Time Intelligence (49–55)

49. What is the purpose of TODAY() and NOW()?
- TODAY() Returns today's date only.
- NOW() Returns the current date and current time.
50. What is the difference between YEAR(), MONTH(), and DAY()?
- From the date column
- year: Returns the year.
- month: Returns the month number.
- day: Returns the day of the month.
51. What is TOTALYTD()? Give an example.
- calculates Year-to-Date (YTD) totals
- YTD Sales =
TOTALYTD(
    SUM(Sales[Sales]),
    'Date'[Date]
)
52. What is TOTALMTD()?
- calculates Month-to-Date totals
- returns the cumulative total from the beginning of the current month up to the selected date.
53. What is TOTALQTD()?
- calculates Quarter-to-Date totals.
- returns the cumulative total from the beginning of the current quarter up to the selected date.
54. What are Time Intelligence functions, and what is required before using them?
- Time Intelligence functions perform calculations over dates
- A Date column is must
55. What is the difference between YTD, MTD, and QTD?
- YTD: Sums your sales all the way from January 1st, 2026 to July 28th, 2026.(year wise)
- MTD:Sums your sales from July 1st, 2026 to July 28th, 2026.(month wise)
- QTD:July is the start of Quarter 3. Therefore, QTD sums your sales from July 1st, 2026 to July 28th, 2026 (in this specific case, it matches MTD because July is the first month of the quarter).
